```python
#Write a python program that, given an input list of any level of complexity/nestedness, will return the inner most list plus 1. 
#This is to be done with a while loop. Note: the input will contain only integers or lists.

my_list = [[3, 7], [12, 5], [8, [4, 9]], [7, 3], [6, [11, [14, 1]]]]
def inner_plus_one(my_list):
    current = my_list                             #current is saying "where am i" and what the list looks like
    while isinstance(current[-1], list):          #isinstance 
        current = current[-1]
    return [x + 1 for x in current]
    
print(inner_plus_one(my_list))
```
