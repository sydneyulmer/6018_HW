```python
#1b:
def corrected_add_function(arg1, arg2):
    arg1_index = 0
    while arg1_index < len(arg1):
        arg_2_sum = 0
        for arg2_elements in arg2:                                                   #for loop repeats it 3 times
            #print(f"Debugging: summing {arg1[arg1_index]} + {arg2[arg1_index]}")
            arg_2_sum = arg1[arg1_index] + arg2[arg1_index]
          
        arg1[arg1_index] = arg_2_sum
        arg1_index += 1
    return arg1


arg1 = [3, 7, 9]
arg2 = [1, 3, 5]
expected_output = [4, 10, 14]
actual_output = corrected_add_function(arg1, arg2)

print(f"The expected answer is: {expected_output}")
print(f"The actual output is: {actual_output}")
```

```python
#2a:
def corrected_add_function_two(arg1, arg2):
   #numeric section
   if sum([type(i)==int for i in arg1])==len(arg1) and \
      sum([type(i)==int for i in arg2])==len(arg2):
         arg1_index=0
         while arg1_index < len(arg1):
            arg_2_sum = 0
            for arg2_elements in arg2:
               arg_2_sum = arg1[arg1_index] + arg2[arg1_index]
            arg1[arg1_index]=arg_2_sum  
            arg1_index+=1
         return arg1
   #string section
   elif sum([type(i)==str for i in arg1])==len(arg1) and \
      sum([type(i)==str for i in arg2])==len(arg2):
         arg1_index=0
         while arg1_index < len(arg1):
            arg_2_sum = ''
            for arg2_elements in arg2:
               arg_2_sum += arg2_elements
            arg1[arg1_index]=arg1[arg1_index]+str(arg_2_sum)
            arg1_index+=1
         return arg1
arg_str_1=['7','10','21']
arg_str_2=['2','2', 2]

corrected_add_function_two(arg_str_1,arg_str_2)
```
